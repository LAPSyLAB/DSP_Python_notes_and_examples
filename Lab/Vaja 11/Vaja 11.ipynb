## Načrtovanje NEO filtrov
Diferenčna enačba: $\sum_{k=0}^{N}a_ky[n-k]=\sum_{k=0}^{M}b_ky[n-k]$

Prenosna funkcija: $H(z)=\frac{Y(z)}{X(z)}=\frac{\sum_{k=0}^{M}b_kz^{-k}}{\sum_{k=0}^{N}a_kz^{-k}}$

Prenosna funkcija: $H(z)=(\frac{b_0}{a_0})\frac{\prod_{k=0}^{M}(1-n_kz^{-1})}{\prod_{k=0}^{N}(1-p_kz^{-1})}$

Amplitudni odziv: $|H(e^{j\omega})|=(\frac{b_0}{a_0})\frac{\prod_{k=0}^{M}|e^{j\omega}-n_k|}{\prod_{k=0}^{N}|e^{j\omega}-p_k)}$

$H(z)$ je ulomek dveh polinomov ($H(z)=\frac{P(z)}{Q(z)}$). Ničle $P(z)$ so ničle sistema, ničle $Q(z)$ so poli sistema.

**Opomba**: Ker so koeficienti polinomov $P(z)$ in $Q(z)$ realni, so poli in ničle bodisi realne, ali pa nastopajo v kompleksno konjugiranih parih. 

Frekvenčni odziv je $H(z)$ ovrednoten na enotski krožnici v Z-ravnin:
- Poli zvišujejo vrednost frekvenčnega odziva (odvisno od bližine krožnice).
- Ničle zmanjšujejo vrednost frekv. odziva (odvisno od bližine krožnice).

Nekaj uporabnih funkcij iz knjižnice Scipy:
- Pretvorba ničel, polov in ojačanja v prenosno fukcijo ```b, a = sgn.zpk2tf(z, p, K)```.
- Pretvorba prenosne fukcije v ničle, pole in ojačanje ```z, k, p = sgn.tf2zkp(b, a)```.
